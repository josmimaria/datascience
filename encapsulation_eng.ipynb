```python
class car:

    def __init__(self , year , make , model ,speed ) :
        self.__year = year
        self.__make = make
        self.__model = model
        self.__speed = 0

    def set_speed(self , speed) :
        self.__speed = 0 if speed < 0 else speed

    def get_speed(self) :
        return self.__speed


```

```python
obj_car = car(1991,"toyota","Innova",65)
```

```python
obj_car.get_speed()
```

```python
obj_car.set_speed(-1234)
```

```python
obj_car.set_speed(3453)
```

```python
class bank_acount:

    def __init__(self , balance ):
        self.__balance = balance

    def deposit(self , amount ) :
        self.__balance = self.__balance + amount

    def withdraw(self , amount) :
        if self.__balance >= amount :
            self.__balance = self.__balance -amount
            return True
        else :
            return False

    def get_balance(self) :
        return self.__balance
```

```python
obj_bank_account = bank_acount(1000)
```

```python
obj_bank_account.get_balance()
```

```python
obj_bank_account.deposit(6000)
```

```python
obj_bank_account.get_balance()
```

```python
obj_bank_account.withdraw(10000)
```

```python
obj_bank_account.withdraw(2000)
```

```python
obj_bank_account.get_balance()
```
